```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = sns.load_dataset("titanic")
df = df.drop_duplicates()
df["age"] = df["age"].fillna(df["age"].median())
df = df.drop(columns=["deck"], errors="ignore")
df = df.dropna()

print(df.shape)
df.head()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="sex", y="survived", color="steelblue")
plt.title("Female passengers had a much higher survival rate than males", fontsize=13)
plt.xlabel("Gender")
plt.ylabel("Survival rate")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="pclass", y="survived", color="teal")
plt.title("1st class passengers had the highest survival rate", fontsize=13)
plt.xlabel("Ticket class")
\
plt.ylabel("Survival rate")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="pclass", y="survived", hue="sex")
plt.title("Females had a higher survival rate than males in every class", fontsize=13)
plt.xlabel("Ticket class")
plt.ylabel("Survival rate")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="age", hue="survived", bins=30, kde=True)
plt.title("Young children had better chances of survival", fontsize=13)
plt.xlabel("Age")
plt.ylabel("Number of passengers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="survived", y="fare")
plt.title("Survivors paid a higher average fare", fontsize=13)
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Fare")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle("Titanic Survival Analysis", fontsize=18, fontweight="bold")

sns.barplot(data=df, x="sex", y="survived", color="steelblue", ax=axes[0, 0])
axes[0, 0].set_title("Females had higher survival rate")

sns.barplot(data=df, x="pclass", y="survived", color="teal", ax=axes[0, 1])
axes[0, 1].set_title("1st class had highest survival")

sns.histplot(data=df, x="age", hue="survived", bins=30, ax=axes[1, 0])
axes[1, 0].set_title("Children survived more")

sns.boxplot(data=df, x="survived", y="fare", ax=axes[1, 1])
axes[1, 1].set_title("Survivors paid higher fare")

plt.tight_layout()
plt.savefig("titanic_dashboard.png", dpi=150)
plt.show()
```

## Data Story

Problem: What factors affected a passenger's chances of survival on the Titanic?

Findings: Female passengers, 1st class passengers, young children, and passengers who paid higher fares had noticeably better survival rates.

Reason: The "women and children first" policy during evacuation, along with better cabin locations for 1st class passengers, likely explain these patterns.

Conclusion: Survival on the Titanic was not random. Gender, ticket class, age and fare all played a significant role, as confirmed by the hypothesis tests in the EDA.
